## From speed to power

```python
def calc_turbine_speed(speed, turbine_height=70, data_height=10, hellman_exponent=1/7):
    turbine_speed = speed * (turbine_height / data_height) ** hellman_exponent
    return turbine_speed

# Discrete power curve: for wind speeds 0 to 24 m/s
power_curve = np.array([
    0, 0, 0, 102, 257, 471, 750, 1194, 1713, 2227, 2465, 2500,
    2500, 2500, 2500, 2500, 2500, 2500, 2500, 2500, 2500, 2500, 2500, 2400, 2250
])

wind_speeds = np.arange(0, 25)

def speed_to_power(speed_array):
    speed_array = np.clip(speed_array, 0, 24)  # avoid out-of-bounds
    return np.interp(speed_array, wind_speeds, power_curve)
```

```python
# use the function above we can firstly calculate turbine speed and then the power 
# working with vector
gdf['turbine_speed'] = gdf['speed'].apply(calc_turbine_speed)
gdf['power'] = gdf['speed'].apply(speed_to_power)
```

## Net present value (NPV)
<img src="https://github.com/BENVYZ/practical_images/blob/main/BENV0093/NPV.png?raw=true" width="500">

```python
def calc_NPV(annual_revenue, lifetime_yrs, CAPEX, dis=0.05, OPEX=0):

    t = np.arange(1, lifetime_yrs + 1)
    discounted_cash_flows = (annual_revenue - OPEX) / (1 + dis) ** t #annual netflow
    NPV = discounted_cash_flows.sum() - CAPEX
    print(discounted_cash_flows.sum())
    return round(NPV, 0)
```

```python
#e.g., one power station with installed capacity 500MW, selling price is 103 USD/MWh,  Discount rate = 5%, 
# Lifetime = 25 years and CapEx = 1.6 billion USD

revenue=500*8760*0.35*103 #the annual power generation 
life_span = 25
CAPEX = 1.6*10**9 ##1.6 billion


calc_NPV (revenue,life_span,CAPEX ,dis) 
```

## Levelized cost of energy (LCOE)
LCOE stands for Levelized Cost of Electricity, a metric used to measure the average cost to build and operate a power plant over its lifetime, expressed as a cost per unit of electricity generated (usually USD/MWh). It's a way to compare the cost-effectiveness of different energy generation methods by accounting for both initial and ongoing expenses over a project's full lifespan. 

<img src="https://github.com/BENVYZ/practical_images/blob/main/BENV0093/LCOE.png?raw=true" width="500">

```python
def calc_LCOE(capex, opex, annual_generation, lifetime, discount_rate):

    years = np.arange(1, lifetime + 1)

    # discount factors
    discount_factors = 1 / (1 + discount_rate) ** years

    # discounted totals
    total_discounted_cost = capex + np.sum(opex * discount_factors)
    total_discounted_generation = np.sum(annual_generation * discount_factors)
    
    lcoe = total_discounted_cost / total_discounted_generation
    return round(lcoe, 2)
```

```python
opex =0
lifetime = 25
capex = 1.6*10**9 ##1.6 billion
annual_generation = 500*8760*0.35
discount_rate = 0.05 

calc_LCOE(capex, opex, annual_generation, lifetime, discount_rate)

# 74 USD/MWh
```
